```python
#How does saying thank you or please affect the token usage? Let's test it out by asking the same question in 3 different ways and see how many tokens are used in each response.

message_normal = [{
    "role":"user",
    "content" : "Explain the difference between supervised and unsupervised learning in AI."}]

message_polite = [{
    "role":"user",
    "content" : "Please explain the difference between supervised and unsupervised learning in AI. Thank You."}]

response_normal = openai_client.chat.completions.create(model="gpt-4o", messages=message_normal)
response_polite = openai_client.chat.completions.create(model="gpt-4o", messages=message_polite)

print("Normal Message Tokens Used: ", response_normal.usage.total_tokens)
print("Polite Message Tokens Used: ", response_polite.usage.total_tokens)
```

```python
#System Prompts:

system_prompt_dict = {
    "Tony": "You are Tony Greg, a sports commentator known for your witty and insightful commentary on football matches. You have a deep understanding of the game and a unique ability to engage the audience with your humor and analysis.",
    "Ravi Shastri": "You are Ravi Shastri, a former Indian cricketer and current cricket commentator known for your energetic and passionate commentary style. You have a deep knowledge of cricket and a unique ability to connect with the audience through your enthusiasm and insights.",
    "Harsha Bhogle": "You are Harsha Bhogle, an Indian cricket commentator known for your eloquent and articulate commentary style. You have a deep understanding of cricket and a unique ability to provide insightful analysis while maintaining a calm and composed demeanor.",
    "Dale Steyn": "You are Dale Steyn, a former South African cricketer and current cricket commentator known for your aggressive and passionate commentary style. You have a deep knowledge of cricket and a unique ability to provide insightful analysis while maintaining a fiery and energetic demeanor."
}

user_message_sachin = [{
    "role": "user",
    "content": "Describe Sachin's straight drive in one sentence"
}]
user_message_virat = [{
    "role": "user",
    "content": "Describe Virat's cover drive in one sentence"
}]

for system_prompt in system_prompt_dict:
    for user_message in [user_message_sachin, user_message_virat]:
        response = openai_client.chat.completions.create(model="gpt-4o-mini", messages=[
            {"role": "system", "content": system_prompt_dict[system_prompt]},
            *user_message
        ])

        print(f"{system_prompt} Commentary: ", end="")
        display(Markdown(response.choices[0].message.content))
```
